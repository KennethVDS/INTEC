**1-Schrijf een functie die het volume van een bol (sphere) berekent op basis van zijn straal (radius).**
<p>Het volume van een bol wordt gegeven als $$\frac{4}{3} πr^3$$</p>

```python
import math
def vol(rad):
    volume = (4/3) * math.pi * (rad ** 3)
    print(f"The volume of the sphere with radius {rad} is {volume:.2f}")
    return volume



```

```python
# Check
vol(2)
```

**2-Schrijf een functie die controleert of een getal binnen een bepaald bereik valt (inclusief hoog en laag)**

```python
def ran_check(num,low,high):
    while low <= high:
        if num >= low and num <= high:
            return f"{num} is between {low} and {high}."
        elif num < low:
            return f"{num} is lower than {low}."
        else: 
            return f"{num} is higher than {high}."
    return "Check your input."
    
```

```python
# Controleren
ran_check(10,3,100)
```

Als je alleen een boolean wilt retourneren:

```python
def ran_bool(num,low,high):
    while low <= high:
        if num >= low and num <= high:
            return True
        else: 
            return False
    return "Check your False input."
```

```python
ran_bool(7,5,10)
```

**3-Schrijf een Python-functie die een string accepteert en het aantal hoofdletters en kleine letters berekent.**

    Sample String : 'Hello Mr. Rogers, how are you this fine Tuesday?'
    Expected Output : 
    No. of Upper case characters : 4
    No. of Lower case Characters : 33

HINT: Twee string-methoden die nuttig kunnen zijn: **.isupper()** en **.islower()**

Als je ambitieus bent, verken dan de module Collecties om dit probleem op te lossen!

```python
def up_low(s):
    up_count = 0
    low_count = 0
    for i in s:
        if i.isupper():
            up_count += 1
        elif i.islower():
            low_count += 1
    print(f"No. of upper case Characters : {up_count}")
    print(f"No. of Lower case Characters : {low_count}")
```

```python
s = 'Hello Mr. Rogervvvvvvvvvvs, how are yOdKNJHDFBJOQAZu this fine Tuesday?'
up_low(s)
```

**4-Schrijf een Python-functie die een lijst nodig heeft en een nieuwe lijst retourneert met unieke elementen van de eerste lijst.**

    Sample List : [1,1,1,1,2,2,3,3,3,3,4,5]
    Unique List : [1, 2, 3, 4, 5]

```python
def unique_list(lst):
    unique_list = list(set(lst))
    return unique_list
```

```python
unique_list([1,1,1,1,2,2,3,3,3,3,4,5])
```

**5-Schrijf een Python-functie om alle getallen in een lijst te vermenigvuldigen (multiply).**

    Sample List : [1, 2, 3, -4]
    Expected Output : -24

```python
from functools import reduce
from operator import mul
def multiply(numbers):  
    result = reduce(mul, numbers)
    return result
```

```python
multiply([-7,28,3,120])
```

**6-Schrijf een Python-functie die controleert of een woord of zin palindroom is of niet.**

Opmerking: een palindroom is een woord, zin of reeks die achterstevoren hetzelfde leest als vooruit, bijvoorbeeld "madam", "kayak", "racecar" of een zinsnede "nurses run". Hint: misschien wil je de methode .replace() in een string bekijken om te helpen bij het omgaan met spaties. Zoek ook op Google hoe u een string in Python kunt omkeren, er zijn enkele slimme manieren om dit te doen met slicing-notatie.

```python
def palindrome(s):
    if s == s[::-1]:
        return True
    else: return False
```

```python
palindrome('hellep')
```

**7-Schrijf een Python-functie om te controleren of een string pangram is of niet. (Stel dat de ingevoerde tekenreeks geen leestekens/punctuation heeft)**

     Let op: Pangrammen zijn woorden of zinnen die elke letter van het alfabet minstens één keer bevatten.
     Bijvoorbeeld: "The quick brown fox jumps over the lazy dog"

Hint: misschien wilt u de methode .replace() gebruiken om spaties te verwijderen.

Hint: kijk naar de [string-module](https://stackoverflow.com/questions/16060899/alphabet-range-in-python)

Hint: als je [set-vergelijkingen](https://medium.com/better-programming/a-visual-guide-to-set-comparisons-in-python-6ab7edb9ec41) wilt gebruiken

```python
import string

def is_pangram(str1, alphabet=string.ascii_lowercase):
    str1 = str1.lower()
    str1 = str1.replace(" ", "")
    if set(str1) == set(alphabet):
        return True
    else: return False
```

```python
is_pangram("The quick brown fox jumps over the lazy dog")
```
