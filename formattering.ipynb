```python
print('{0:8} | {1:9}'.format('Fruit', 'Quantity'))
print('{0:5} | {1:10}'.format('Apples', 3.))
print('{0:7} | {1:2}'.format('Oranges', 10))
```

```python
print('{0:<8} | {1:^8} | {2:>15}'.format('Left','Center','Right'))
print('{0:<15} | {1:^8} | {2:>8}'.format(11,22,33))
```

```python
print('{0:=<8} | {1:-^8} | {2:.>8}'.format('Left','Center','Right'))
print('{0:=<8} | {1:-^8} | {2:,>22}'.format(11,22,33))
```

```python
print('This is my ten-character, two-decimal number:%10.2f' %13.579)
print('This is my ten-character, two-decimal number:{0:10.2f}'.format(13.579))
```

```python
name = 'Jackie'

print(f"He said her name is {name}.")
print(f"He said her name is {name!r}")
```

```python
num = 23.456789
print("My 10 character, four decimal number is:{0:10.5f}".format(num))
print(f"My 10 character, four decimal number is:{num:{10}.{6}}")
```
