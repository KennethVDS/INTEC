```python
class Car:
        
        def __init__(self, color, mileage):
            self.color = color
            self.mileage = mileage
            
        def __str__(self):
            return f"The {self.color} car has {self.mileage} miles."
        def drive(self,miles):
            new_mileage = self.mileage + miles
            return new_mileage
        
red_car = Car("red", 20000)
green_car = Car("green", 30000)
print(red_car)
print(green_car)
```

```python
import math
class Line:
    def __init__(self, coor1, coor2):
        self.coor1 = coor1
        self.coor2 = coor2
   
    def distance(self):
        # formula: distance = sqrt((x2 - x1)^2 + (y2 - y1)^2)
        x1, y1 = self.coor1
        x2, y2 = self.coor2
        return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
   
    def slope(self):
        # formula: slope = (y2 - y1) / (x2 - x1)
        x1, y1 = self.coor1
        x2, y2 = self.coor2
        return (y2 - y1) / (x2 - x1)
```

```python
# EEN VOORBEELD SCENARIO OM UIT TE VOEREN

coordinate1 = (3,2)
coordinate2 = (8,10)
li = Line(coordinate1,coordinate2)
```

```python
li.distance()
```

```python
li.slope()
```

6-Vul de klas in

```python
class Cylinder:
    
    def __init__(self,height=1,radius=1):
        self.height = height
        self.radius = radius
        
    def volume(self):
        return self.height * 3.14 * (self.radius)**2
    
    def surface_area(self):
        top = 3.14 * (self.radius)**2
        return (2*top) + (2*3.14*self.radius*self.height)
```

```python
# EEN VOORBEELD SCENARIO OM UIT TE VOEREN

c = Cylinder(2,3)
```

```python
c.volume()
```

```python
#uitvoer
56.52
```

```python
c.surface_area()
```
