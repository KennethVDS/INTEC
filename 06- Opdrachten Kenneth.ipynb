1- Programma Uitleg:

1. Er wordt een klasse met de naam Calculator gemaakt en de methode __init__() wordt gebruikt om de waarden van die klasse te initialiseren.
2. Methoden voor het optellen, aftrekken, vermenigvuldigen, delen van twee getallen en het retourneren van hun respectievelijke resultaten zijn gedefinieerd.
3. Het menu wordt afgedrukt en de keuze wordt gemaakt door de gebruiker.
4. Er wordt een object voor de klasse gemaakt met de twee nummers van de gebruiker die als parameters worden doorgegeven.
5. Met behulp van het object wordt de respectieve methode aangeroepen volgens de keuze van de gebruiker.
6. Als de keuze 0 is, wordt de lus afgesloten.
7. Het eindresultaat wordt afgedrukt.

````python
class Calculator():
    def __init__(self, a, b):
        self.num1 = a
        self.num2 = b
        
    def add(self):
        solution = self.num1 + self.num2
        return solution
    
    def sub(self):
        solution = self.num1 - self.num2
        return solution
    
    def mul(self):
        solution = self.num1 * self.num2
        return solution
    
    def div(self):
        solution = self.num1 / self.num2
        return solution

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

obj = Calculator(a, b)

choice = 1

while choice != 0:
    print("0. Exit")
    print("1. Add")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")

    choice = int(input("Enter choice: "))
    if choice == 1:
        obj.add()
        print("Result:", obj.add())
        break
    elif choice == 2:
        obj.sub()
        print("Result:", obj.sub())
        break
    elif choice == 3:
        obj.mul()
        print("Result:", obj.mul())
        break
    elif choice == 4:
        obj.div()
        print("Result:", obj.div())
        break
    elif choice == 0:
        print("Exiting!")
        break
    else:
        break

print()
````

2.1. Definieer een class genaamd Dog met een constructor __init__ die eigenschappen name en age initialiseren. De class moet ook een methode __str__ hebben die een leesbare string retourneert met de naam en leeftijd van de hond. Daarnaast moet er een methode speak zijn die een geluid teruggeeft, gebaseerd op hetgeen is opgegeven.
 
2.2. Creëer een subklasse genaamd GoldenRetriever die is afgeleid van de Dog klasse. De GoldenRetriever klasse moet een methode speak hebben die, als er geen geluid wordt opgegeven, het standaard geluid "Bark" teruggeeft. Maak gebruik van de super() functie om de speak methode van de ouderklasse te gebruiken.

````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def __str__(self):
        return f"Name: {self.name}, Age: {self.age}"
        
    def speak(self, sound):
        return f"{self.name} says {sound}" 
````

````python
class GoldenRetriever(Dog):
    def speak(self, sound="Bark"):
        return super().speak(sound)
    
    
dog = GoldenRetriever('Golden Retriever', 5)
````

3.1. Definieer een class genaamd Rectangle met een constructor __init__ die lengte en breedte initialiseert. Deze class moet een methode area hebben die de oppervlakte van het rechthoekige object berekent.
 
3.2. Creëer een subklasse genaamd Square die is afgeleid van de Rectangle klasse. Gebruik de super() functie in de __init__ methode om de lengte en breedte in te stellen op basis van de zijde van het vierkant.
 
3.3. Maak een instantie van de Square klasse met zijdelengte 4 en druk de oppervlakte ervan af.

````python
class Rectangle:
    def __init__(self, lengte, breedte):
        self.lengte = lengte
        self.breedte = breedte
        
    def oppervlakte(self):
        return self.lengte * self.breedte
        
class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)
        
# Maak een instantie van de Square klasse met zijdelengte 4
vierkant = Square(25)

# Druk de oppervlakte ervan af
print(f"De oppervlakte is: {vierkant.oppervlakte()}")
````
