**Gebruik <code>for</code>, .split() en <code>if</code> om een instructie te maken die woorden afdrukt die beginnen met 's':**

```python
st = 'Print only the words that start with s in this sentence'
```

```python
for word in st.split():
    if word[0].lower() == 's':
        print(word)

   
```

**Gebruik range() om alle even getallen van 0 tot 10 af te drukken.**

```python

for num in range(0,10):
    if num % 2 == 0:
        print(num)
```

*Gebruik Lijstbegrip om een lijst te maken van alle getallen tussen 1 en 50 die deelbaar zijn door 3.**

```python
#Code in this cell
listje=[]

for i in range(1, 50):
    if (i%3==0):
        listje.append(str(i))
print(','.join(listje))
```

**Ga door de onderstaande reeks en als de lengte van een woord gelijk is, drukt u "even!" af**

```python
st = 'Print every word in this sentence that has an even number of letters'
```

```python
#Code in this cell
for word in st.split():
    if len(word)%2 == 0:
        print(word)
        
```

**Schrijf een programma dat de gehele getallen van 1 tot 100 afdrukt. Maar voor veelvouden van drie drukt u "Fizz" af in plaats van het getal, en voor veelvouden van vijf drukt u "Buzz" af. Voor getallen die een veelvoud zijn van zowel drie als vijf, drukt u "FizzBuzz" af.**

```python

#Code in this cell
for i in range(1,101):
    if(i%2==0):
        print(i)
        if(i%3==0):
            print('Fizz')
        if(i%5==0):
            print('Buzz')
        if(i%3==0 and i%5==0):
            print('FizzBuzz')
            
```

**Gebruik Lijstbegrip om een lijst te maken van de eerste letters van elk woord in de onderstaande reeks:**

```python
st = 'Create a list of the first letters of every word in this string'
```

```python
#Code in this cell
listje=[]
for word in st.split():
    listje.append(word[0])
print(','.join(listje))
```
